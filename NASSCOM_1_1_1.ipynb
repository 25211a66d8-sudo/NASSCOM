### 1. Basic Python Statement and Indentation

```python
print("Hello, Python!")  # This is a comment

if True:
    print("This is indented correctly.")
else:
    print("This won't run.")
```

### 2. Variables and Dynamic Typing

```python
name = "Alice"
age = 25
height = 5.6

print("Name:", name)
print("Age:", age)
print("Height:", height)

name = 123
print("New value of name:", name)
```

### 3. Data Types: Integer and Float

```python
a = 10
print(type(a))

b = 3.14
print(type(b))
```

### 4. Other Data Types: String, Boolean, List, and Dictionary

```python
c = "Python"
print(type(c))

d = True
print(type(d))

e = [1, 2, 3, "four"]
print(type(e), e)

f = {"name": "Alice", "age": 25}
print(type(f), f)
```

### 5. Output and Input

```python
print("Welcome to Python I/O!")

user_name = input("Enter your name: ")
print("Hello,", user_name)
```

### 6. Type Casting Input

```python
user_age = int(input("Enter your age: "))
print("You will be", user_age + 1, "next year.")
```
